```python
def search(x):
    lst = [2,3,4,5,6,7,8,9]
    left = 0
    right = 0
    mid = len(lst)/2
    if x >= mid:
        left+=1
        print(lst[x])
        
        if x == left:
            print(f"{x} found {lst[x]}")
    elif x <= mid:
        right+=1
        if x == right:
            print(f"{x} found {lst[x]}")

search(5)
    

```
