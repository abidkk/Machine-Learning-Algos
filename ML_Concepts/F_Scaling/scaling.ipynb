# FEATURE SCALING IN MACHINE LEARNING

###  What is Feature Scaling?
Feature scaling is the process of transforming or balancing the values of your features (the variables you use for your model) to a similar range. It essentially adjusts the scale of your data so that features with larger values don't dominate the model and vice versa.

### Common Feature Scaling Techniques
1. Standardization (Z-score Scaling)
2. Min-Max Scaling (Normalization)
3.	Robust Scaling
4.	Max Absolute Scaling

```python
import pandas as pd

data = {
    "Name": ["Adam", "Neha", "Zoha", "John", "Zahid", "Basit", "Adeel"],
    "Age": [21,31,12,44,63,28, 25],
    "Salary" :[31000, 55000, 70000, 65000, 55000, 45000, 40000]

}
df = pd.DataFrame(data)
df
```

```python
df["Salary"]
```

---

## 1. Standard Scaling
#### 1.	Standardization (Z-score Scaling): 
- a.	How it works: Transforms features to have a mean of 0 and a standard deviation of 1. x' = (x - mean) / std
- b.	Use Case: Often preferred because it doesn't require you to define a specific range. It focuses on the position of the data points, making it more robust to outliers.
- c.	Example: Commonly used for continuous features like age, income, or height.

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
ss.fit(df[["Salary"]])
ss.transform(df[["Salary"]])
```

```python
new_df = pd.DataFrame(ss.transform(df[["Salary"]]))
new_df
```

```python
df["Salary_1"] = new_df[0]
df
```

---

## 2. Normalization Scaling : (Min, Max)
#### 2.	Min-Max Scaling (Normalization): 
- a.	How it works: Scales features to a range between 0 and 1. x' = (x - min) / (max - min)
- b.	Use Case: When you want to preserve the original distribution of the data, and the data is relatively bounded. Good for algorithms that expect a particular range of values.
- c.	Example: Normalization is often used for images (pixel values) or data with a range of values.

```python
from sklearn.preprocessing import MinMaxScaler
ms = MinMaxScaler()
ms.fit(df[["Salary"]])
# ms.transform(df[["Salary"]])
new_df2 = pd.DataFrame(ms.transform(df[["Salary"]]))
new_df2
```

```python
df["Salary_2"] = new_df2[0]
df
```

---


## 3. Robast Scaler


- a.	How it works: Scales features using the median and interquartile range (IQR). This is more robust to outliers than min-max scaling.
- b.	Use Case: When dealing with datasets that contain outliers that can significantly distort model results.


```python
from sklearn.preprocessing import RobustScaler
rs = RobustScaler()
rs.fit(df[["Salary"]])
# ss.transform(df[["Salary"]])
new_df3 = pd.DataFrame(rs.transform(df[["Salary"]]))
df["Salary_3"] = new_df3[0]
df
```

## 4.	Max Absolute Scaling

- a.	How it works: Scales features by taking the absolute value of each element, and then dividing by the maximum absolute value.
- b.	Use Case: Similar to min-max but often works better with non-negative values, as it avoids artificially shrinking the range of values.


```python
from sklearn.preprocessing import MaxAbsScaler
mas = MaxAbsScaler()
mas.fit(df[["Salary"]])
# ss.transform(df[["Salary"]])
new_df4 = pd.DataFrame(mas.transform(df[["Salary"]]))
df["Salary_4"] = new_df4[0]
df
```
