```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
data = {

    "Area":[350,600,240,700,540,300, 500, 640, 430, 560],
    "Bedrooms":[2,3,2,4,3,2,3,4,3,3],
    "Price": [20,30,13,40,25, 28, 29, 40, 32,38]
}

print(data)
```

```python

df = pd.DataFrame(data)
df
```

```python
x = df[["Area", "Bedrooms"]]
y = df["Price"]


print(f"x: {x}")
print(f"y: {y}")
```

```python
model = LinearRegression()
model.fit(x,y)

coef_area= model.coef_[0]
coef_beds= model.coef_[1]
intercept = model.intercept_

print(coef_area)
print(coef_beds)
print(intercept)
```

```python
plt.scatter(x["Area"], y, label="Actual")

plt.plot(
    x["Area"],
    model.predict(x),
    color="red",
    label="Predicted"
)

plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Price vs Area (Bedrooms included in model)")
plt.legend()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Area vs Price
axes[0].scatter(x["Area"], y)
axes[0].plot(x["Area"], model.predict(x))
axes[0].set_xlabel("Area")
axes[0].set_ylabel("Price")

# Bedrooms vs Price
axes[1].scatter(x["Bedrooms"], y)
axes[1].plot(x["Bedrooms"], model.predict(x))
axes[1].set_xlabel("Bedrooms")
axes[1].set_ylabel("Price")

plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

ax.scatter(x["Area"], x["Bedrooms"], y)

ax.set_xlabel("Area")
ax.set_ylabel("Bedrooms")
ax.set_zlabel("Price")

plt.show()
```

```python
residuals = y - model.predict(x)

plt.scatter(x, residuals)
plt.axhline(0)
plt.xlabel("Area")
plt.ylabel("Residual")
plt.title("Residual Plot")
plt.show()
```

```python
X = df[["Area", "Bedrooms"]]
y = df["Price"]
y_pred = model.predict(X)
```

```python
residuals = y - y_pred

plt.scatter(y_pred, residuals)
plt.axhline(y=0)
plt.xlabel("Predicted Price")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
```

```python
plt.scatter(X["Area"], residuals)
plt.axhline(0)
plt.xlabel("Area")
plt.ylabel("Residuals")
plt.title("Residuals vs Area")
plt.show()
```

```python
plt.scatter(X["Bedrooms"], residuals)
plt.axhline(0)
plt.xlabel("Bedrooms")
plt.ylabel("Residuals")
plt.title("Residuals vs Bedrooms")
plt.show()
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y, y_pred)
print("MSE:", mse)
```

```python
rmse = mse ** 0.5
print("RMSE:", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, y_pred)
print("R² Score:", r2)
```
