# ENCODING 
## What is Encoding?
At its core, encoding is about representing data using a specific system. This system ensures that the data can be correctly interpreted by the algorithm or system you’re using. Without encoding, your data could be meaningless or lead to incorrect results.

## Why is Encoding Important?
Machine Learning Algorithms: Many machine learning algorithms require numerical input. Data needs to be transformed into a numerical format (e.g., 0 to 1, or decimals) before they can be used in calculations.
Data Consistency: Encoding ensures consistent representation of data across different sources or systems.
Improved Performance: Properly encoding data can dramatically improve the performance of machine learning models.

5. Encoding (for categorical data)

- Techniques:

    3. Binary Encoding (Yes/No → 1/0).
