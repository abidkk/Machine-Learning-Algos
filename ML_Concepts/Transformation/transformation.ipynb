# FUNCTION TRANSFORMATION

#### What are Function Transformations?

Simply put, they’re ways to change the mathematical representation of your data. They’re not just about cleaning data; they’re about reshaping the data to enhance the learning process of models.

#### Why Use Them?

- Improved Model Performance: Many algorithms (like linear regression, neural networks) perform better when the data is transformed into a suitable range or format.
- Linearization: Transforms data to a linear space, simplifying calculations and making it easier for models to find patterns.
- Feature Extraction: Identifies important underlying features by modifying the data.
Dimensionality Reduction: Reduce the number of features while preserving important information.


#### When to Use Which Transformation?

1. Log Transformation: For skewed data.
2. Square Root Transformation: For data that's too far off the range.
3. Power Transformation: For non-linear relationships.
4. Normalization: When you want to bring values to a specific scale.

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


data = {
    "Name": ["Adam", "Neha", "Zoha", "John", "Zahid", "Basit", "Adeel"],
    "Age": [21,31,12,44,63,28, 25],
    "Salary" :[31000, 55000, 70000, 65000, 5000, 45000, 400000]
}
df = pd.DataFrame(data)
df
```

```python
df["Salary"]
```

### Transformation

```python
from sklearn.preprocessing import FunctionTransformer
ft = FunctionTransformer(func=np.log1p)
ft.fit(df[["Salary"]])
new_df = pd.DataFrame(ft.transform(df[["Salary"]]))
new_df
```

```python
df["Salary_1"] = new_df["Salary"]
df
```

## Ploting Before & After Transformation

```python
plt.figure(figsize=(12, 4)) 

plt.subplot(1, 2, 1)  # Create the first subplot
sns.distplot(df["Salary"])
plt.title("BEFORE TRANSFORMATION")


plt.subplot(1, 2, 2)  # Create the second subplot
sns.distplot(df["Salary_1"])
plt.title("AFTER TRANSFORMATION")
plt.show()
```
