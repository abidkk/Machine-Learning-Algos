```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = {
    "Name": ["Adam", "Neha", "Zoha", "John", "Zahid", "Basit", "Adeel"],
    "Age": [21,840,12,44,63,310, 55]
}
data
```

```python
df = pd.DataFrame(data)
df
```

## Checking Outliers for Age Column
- Note: before moving to outliers make sure handle missing or duplicate values

### Box Plot

```python
plt.figure(figsize=(8,2))
sns.boxplot(x="Age",data=df)
plt.show()
```

### Disribution Plot

```python
plt.figure(figsize=(12,8))
sns.displot(df["Age"], kde=True)
plt.show()
```

```python
q1 = df["Age"].quantile(0.25)
q2 = df["Age"].quantile(0.50)
q3 = df["Age"].quantile(0.75)

q1, q2, q3
```

```python
iqr = q3-q1
iqr
```

```python
max_range = q1 + (1.5 * iqr)
max_range
```

```python
min_range = q1 - (1.5 * iqr)
min_range
```

```python
newAge = df[df["Age"]<= max_range]
newAge
```

```python
plt.figure(figsize=(8,2))
sns.boxplot(x="Age",data=newAge)
plt.show()
```
