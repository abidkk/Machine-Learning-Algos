```python
brand_mode = new_df.Brand.mode()
brand_mode
```

```python
brand_mode = new_df.Brand.mode()
brand_mode
new_df.Brand = new_df.Brand.fillna(brand_mode)
```

```python
new_df.Brand.isnull().sum()
```

```python
condition_mode = new_df.Condition.mode()
condition_mode
new_df.Condition = new_df.Condition.fillna(condition_mode)
```

```python
fuel_mode = new_df.Fuel.mode()
fuel_mode
new_df.Fuel = new_df.Fuel.fillna(fuel_mode)
```

```python
kms_med = new_df.KMs_Driven.median()
kms_med
new_df.KMs_Driven = new_df.KMs_Driven.fillna(kms_med)
```

```python
model_mode = new_df.Model.mode()
model_mode
new_df.Model = new_df.Model.fillna(model_mode)
```

```python
reg_city_mode = new_df.Registered_City.mode()

reg_city_mode
new_df.Registered_City = new_df.Registered_City.fillna(reg_city_mode)
```

```python
year_mode = new_df.Year.mode()
year_mode
new_df.Year = new_df.Year.fillna(year_mode)
```
