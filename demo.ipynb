```python
import pandas as pd
import random

# Define possible values for each column
brands = ["Dell", "HP", "Apple", "Lenovo", "Asus", "Acer", "MSI"]
os_list = ["Windows", "macOS", "Linux"]
cpus = ["Intel i5", "Intel i7", "Intel i9", "AMD Ryzen 5", "AMD Ryzen 7", "Apple M1", "Apple M2"]
gpus = ["Intel Integrated", "NVIDIA GTX 1650", "NVIDIA RTX 3060", "NVIDIA RTX 4060", "AMD Radeon RX 5600M"]
build_quality = ["Poor", "Fair", "Good", "Excellent"]
display_quality = ["HD", "Full HD", "2K", "4K"]
colors = ["Black", "Silver", "Grey", "Blue", "Red"]
warranty_years = [0, 1, 2, 3]

# Function to generate a random price
def generate_price(ram, storage, cpu, gpu, build, display, brand):
    base = 300
    base += ram * 20
    base += storage * 0.4
    if "i7" in cpu or "Ryzen 7" in cpu or "M2" in cpu:
        base += 250
    elif "i9" in cpu:
        base += 400
    if "RTX" in gpu:
        base += 500
    elif "GTX" in gpu:
        base += 200
    elif "Radeon" in gpu:
        base += 300
    if build == "Excellent":
        base += 300
    elif build == "Good":
        base += 150
    if display == "4K":
        base += 400
    elif display == "2K":
        base += 200
    if brand == "Apple":
        base += 500
    return round(base + random.randint(-100, 200), 2)

# Generate dataset
rows = []
for _ in range(50):
    brand = random.choice(brands)
    ram = random.choice([4, 8, 16, 32])
    storage = random.choice([256, 512, 1024, 2048])
    weight = round(random.uniform(1.0, 3.0), 2)
    screen_size = random.choice([13.3, 14.0, 15.6, 17.0])
    battery = random.choice([4, 6, 8, 10, 12])
    os = "macOS" if brand == "Apple" else random.choice(os_list)
    cpu = "Apple M1" if brand == "Apple" else random.choice(cpus)
    gpu = "Apple Integrated" if brand == "Apple" else random.choice(gpus)
    build = random.choice(build_quality)
    display = random.choice(display_quality)
    warranty = random.choice(warranty_years)
    touchscreen = random.choice(["Yes", "No"])
    backlit = random.choice(["Yes", "No"])
    fingerprint = random.choice(["Yes", "No"])
    color = random.choice(colors)
    price = generate_price(ram, storage, cpu, gpu, build, display, brand)

    rows.append([brand, ram, storage, weight, screen_size, battery, os, cpu, gpu, build, display,
                 warranty, touchscreen, backlit, fingerprint, color, price])

# Create DataFrame
columns = ["Brand", "RAM", "Storage", "Weight", "Screen Size", "Battery Life (hours)", 
           "Operating System", "CPU", "GPU", "Build Quality", "Display Quality", 
           "Warranty", "Touchscreen", "Backlit", "Fingerprint", "Color", "Price"]

df = pd.DataFrame(rows, columns=columns)

# Save to CSV
df.to_csv("laptop_prices_sample.csv", index=False)

print("✅ Dataset created and saved as laptop_prices_sample.csv")
print(df.head())
```
