```python
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
data = {
    "Area": [2000,1200,4000,2500,1500,3000,1800],
    "Price":[42000,32000,84000,47000,37000,58000,40000]
}
data
```

```python
df = pd.DataFrame(data)
df
```

```python
x = df["Area"]
y = df["Price"]
```

```python
sns.lmplot(x="Area",y="Price", data=df)
```

```python
reg = LinearRegression()
```

```python
X = x.values.reshape(-1, 1)
reg.fit(X, y)
```

```python
reg.predict([[3800]])
```

### Model Evaluatin

```python
slope = reg.coef_
slope
```

```python
intercept = reg.intercept_
intercept
```
