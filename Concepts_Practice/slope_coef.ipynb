```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
data = {

    "Area":[350,600,240,700,540,300, 500, 640, 430, 560],
    "Price": [20,30,13,40,25, 28, 29, 40, 32,38]
}

print(data)
```

```python

df = pd.DataFrame(data)
df
```

```python
x = df["Area"]
y = df["Price"]


print(f"x: {x}")
print(f"y: {y}")
```

```python
X = x.values.reshape(-1, 1) # converting x 1 dim to 2 dimention bc ML models support multiple features:
# x.ndim
print(X)
print(X.ndim)
```

### 1. Find Slope & Intercept

```python

model = LinearRegression()
model.fit(X, y)

coef = model.coef_[0] # 0 index bc ML supports Mltiple features but in this case its single feature
intercept = model.intercept_

print("Slope/Coeficient: ", coef)
print("Intercept: ", intercept)
```

#### Get r_value (from R²)

```python
r_squared = model.score(X, y)
r_value = np.sqrt(r_squared) if coef > 0 else -np.sqrt(r_squared)

print(r_squared)
print(r_value)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y)
plt.plot(x, model.predict(X))
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Linear Regression Fit")
plt.show()
```

```python
residuals = y - model.predict(X)

plt.scatter(x, residuals)
plt.axhline(0)
plt.xlabel("Area")
plt.ylabel("Residual")
plt.title("Residual Plot")
plt.show()
```
