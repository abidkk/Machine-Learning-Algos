```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": ["Adam", "John", "Soha", "Zara", "Neha", "Laiba", "Zayan", "Abid", "Mariam", "Ahmed"],
    "Gender": ["Male", "Male", "Female", "Female", "Female", "Female", "Male", "Male", "Female", "Male"],
    "Study_Hours": [6, 4, 7, 2, 5, 4, 10, 7, 5, 6],
    "Result":["Pass", "Pass", "Pass", "Fail", "Fail", "Pass", "Pass", "Pass", "Fail", "Pass"]
    # "Grade": ["A", "B", "A", "C", "B", "C", "A", "B", "A", "A"],
    # "City": ["Karachi", "Lahore", "Karachi", "Islamabad", "Karachi","Karachi", "Lahore", "Karachi", "Islamabad", "Karachi"]
}
df = pd.DataFrame(data)
df
```

```python
df["Gender"] = pd.get_dummies(df["Gender"], prefix="Gender", drop_first=True).astype(int)
df
```

```python
df["Result"] = pd.get_dummies(df["Result"], prefix="Result", drop_first=True).astype(int)
df 
```

```python
x = df[["Gender","Study_Hours"]]
# x
y = df[["Result"]]
x 
```

```python
y 
```

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
fs = SequentialFeatureSelector(lr, k_features=2, forward=True)
fs.fit(x,y)
```
